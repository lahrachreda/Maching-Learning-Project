# Preprocessing

## Objective
The objective of this step is to prepare the cleaned dataset for modeling by encoding categorical variables, scaling numerical features, and addressing the severe class imbalance in the target variable.

## Import labraries

```python
import pandas as pd 
from pathlib import Path
```

## Load the dataset 

```python
data =pd.read_csv(Path("../data/raw/fraud_detection_data.csv"))
```

## Separate X and y

```python
X=data.drop(columns=["isFraud","isFlaggedFraud"])
y=data["isFraud"]
```

```python
X.head()
```

## Encoded X and y

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (OneHotEncoder,StandardScaler,LabelEncoder)
cat_cols=[col for col in X.columns if X[col].dtype == 'str']
num_cols=[col for col in X.columns if X[col].dtype in ['int64', 'float64']]
preprocessor =ColumnTransformer(
    transformers=[
        ('num',StandardScaler(),num_cols),
        ('cat',OneHotEncoder(handle_unknown='ignore'),cat_cols)
    ],
    remainder='passthrough')
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)   # fit + transform sur train
y_test_encoded  = le.transform(y_test) 
```
