```python
import joblib
import pandas as pd
from pathlib import Path
loaded_model = joblib.load(
    "../models/model_LR.pkl"
)
X_test=joblib.load(Path("../data/processed/X_test_encoded.csv"))
y_test=joblib.load(Path("../data/processed/y_test_encoded.csv"))
import matplotlib.pyplot as plt
probas = loaded_model.predict_proba(X_test)[:, 1]
plt.hist(probas[y_test==0], bins=50, alpha=0.5, label='non-fraude')
plt.hist(probas[y_test==1], bins=50, alpha=0.5, label='fraude')
plt.legend()
plt.yscale('log')
```

```python
from sklearn.metrics import average_precision_score
print(average_precision_score(y_test, probas))
```
