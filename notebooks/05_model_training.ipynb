# Model Training

## Objective

Train machine learning models for fraud detection and compare their
performance against the baseline model established previously.

## Import labraries

```python
import joblib
from pathlib import Path
from sklearn.metrics import accuracy_score,precision_score,f1_score, confusion_matrix
```

## Import data encoded 

```python
X_train=joblib.load(Path("../data/processed/X_train_encoded.csv"))
X_test=joblib.load(Path("../data/processed/X_test_encoded.csv"))
y_train=joblib.load(Path("../data/processed/y_train_encoded.csv"))
y_test=joblib.load(Path("../data/processed/y_test_encoded.csv"))
```

## Balancing the training data

```python
from imblearn.over_sampling import SMOTE
import pandas as pd

smote = SMOTE(random_state=42)

X_train_balanced, y_train_balanced = smote.fit_resample(X_train,y_train)

print("Avant SMOTE:")
print(pd.Series(y_train).value_counts())

print("\nAprès SMOTE:")
print(pd.Series(y_train_balanced).value_counts())
```

```python
import gc
del X_train
del y_train
gc.collect()
```

## Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
model_LR=LogisticRegression(max_iter=1000,random_state=42)
model_LR.fit(X_train_balanced,y_train_balanced)
y_predict_LR=model_LR.predict(X_test)
print(f"accuracy : {accuracy_score(y_test,y_predict_LR):.4f}")
print("precision: \n",precision_score(y_test,y_predict_LR,average='weighted'))
print("f1 score: \n",f1_score(y_test,y_predict_LR,average='weighted'))
print("confusion matrix: \n",confusion_matrix(y_test,y_predict_LR))
```

### Save the model 

```python
joblib.dump(model_LR, Path("../models/model_LR.pkl"))
```

## Support Vector Machine

```python
from sklearn.svm import SVC
model_SVC=SVC(probability=True,random_state=42)
model_SVC.fit(X_train_balanced,y_train_balanced)
y_predict_SVC=model_SVC.predict(X_test) 
print(f"accuracy : {accuracy_score(y_test,y_predict_SVC):.4f}")
print("precision: \n",precision_score(y_test,y_predict_SVC,average='weighted'))
print("f1 score: \n",f1_score(y_test,y_predict_SVC,average='weighted'))
print("confusion matrix: \n",confusion_matrix(y_test,y_predict_SVC))
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
model_RF=RandomForestClassifier( 
    n_estimators=100,
    random_state=42,
    n_jobs=-1)
model_RF.fit(X_train_balanced, y_train_balanced)
y_predict_RF=model_RF.predict(X_test)
print(f"accuracy : {accuracy_score(y_test,y_predict_RF):.4f}")
print("precision: \n",precision_score(y_test,y_predict_RF,average='weighted'))
print("f1 score: \n",f1_score(y_test,y_predict_RF,average='weighted'))
print("confusion matrix: \n",confusion_matrix(y_test,y_predict_RF))
```

## Xgboost 

```python
from xgboost import XGBClassifier
model_XGB=XGBClassifier(   
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
    )
model_XGB.fit(X_train_balanced, y_train_balanced)
y_predict_XGB=model_XGB.predict(X_test)
print(f"accuracy : {accuracy_score(y_test,y_predict_XGB):.4f}")
print("precision: \n",precision_score(y_test,y_predict_XGB,average='weighted'))
print("f1 score: \n",f1_score(y_test,y_predict_XGB,average='weighted'))
print("confusion matrix: \n",confusion_matrix(y_test,y_predict_XGB))
```

## k_near neighbor


```python
from sklearn.neighbors import KNeighborsClassifier
model_KNN=KNeighborsClassifier(n_neighbors=5,n_jobs=-1)
model_KNN.fit(X_train_balanced,y_train_balanced)
y_predict_KNN=model_KNN.predict(X_test)
print(f"accuracy : {accuracy_score(y_test,y_predict_KNN):.4f}")
print("precision: \n",precision_score(y_test,y_predict_KNN,average='weighted'))
print("f1 score: \n",f1_score(y_test,y_predict_KNN,average='weighted'))
print("confusion matrix: \n",confusion_matrix(y_test,y_predict_KNN))
```

## Decision tree

```python
from sklearn.tree import DecisionTreeClassifier
model_DT=DecisionTreeClassifier(random_state=42)
model_DT.fit(X_train_balanced,y_train_balanced)
y_predict_DT=model_DT.predict(X_test)
print(f"accuracy : {accuracy_score(y_test,y_predict_DT):.4f}")
print("precision: \n",precision_score(y_test,y_predict_DT,average='weighted'))
print("f1 score: \n",f1_score(y_test,y_predict_DT,average='weighted'))
print("confusion matrix: \n",confusion_matrix(y_test,y_predict_DT))
```
