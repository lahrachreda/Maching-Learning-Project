# Baseline

## Objective
Establish a baseline performance for the fraud detection task using a simple model before training more advanced machine learning models. This baseline will serve as a reference for evaluating whether subsequent models provide meaningful improvements.

## Import labraries

```python
import joblib
from pathlib import Path
```

## Import the encoded data

```python
X_train=joblib.load(Path("../data/processed/X_train_encoded.pkl"))
X_test=joblib.load(Path("../data/processed/X_test_encoded.pkl"))
y_train=joblib.load(Path("../data/processed/y_train_encoded.pkl"))
y_test=joblib.load(Path("../data/processed/y_test_encoded.pkl"))
```

## Baseline model 


```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score ,classification_report,confusion_matrix
baseline=DummyClassifier(strategy="most_frequent")
baseline.fit(X_train,y_train)
y_predict_baseline=baseline.predict(X_test)
print(f"accuracy : {accuracy_score(y_test,y_predict_baseline):.4f}")
print("classification report: \n",classification_report(y_test,y_predict_baseline))
print("confusion matrix: \n",confusion_matrix(y_test,y_predict_baseline))
```
